```
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate
```

```
# Task 1: Basic PromptTemplate

template_1 = "Explain {topic} in simple terms for beginners"

prompt_1 = PromptTemplate(
    input_variables=["topic"],
    template=template_1
)

# Test
output_1 = prompt_1.format(topic="Artificial Intelligence")
print(output_1)
```

```
# Task 2: Multiple Inputs

template_2 = "Explain {topic} for {audience} in a {tone} tone"

prompt_2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template_2
)

# Test cases
print(prompt_2.format(topic="AI", audience="beginners", tone="friendly"))
print(prompt_2.format(topic="Python", audience="kids", tone="fun"))
print(prompt_2.format(topic="Deep Learning", audience="engineers", tone="technical"))
```

```
# Task 3: Multiple Templates

teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}"
)

story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

print(teaching_template.format(topic=topic))
print(interview_template.format(topic=topic))
print(story_template.format(topic=topic))
```

```
# Task 4: Chat Prompt System

def get_chat_prompt(role, topic):

    if role == "teacher":
        system_msg = "You are a helpful teacher."
    elif role == "interviewer":
        system_msg = "You are an interviewer."
    elif role == "motivator":
        system_msg = "You are a motivational speaker."
    else:
        system_msg = "You are a helpful assistant."

    chat_prompt = ChatPromptTemplate.from_messages([
        ("system", system_msg),
        ("human", "Talk about {topic}")
    ])

    return chat_prompt.format_messages(topic=topic)


# Test
messages = get_chat_prompt("teacher", "Neural Networks")

for msg in messages:
    print(msg)
```

```
# Task 5: Validation Function

def validate_inputs(audience, tone):

    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        print("Invalid audience! Default set to 'beginner'")
        audience = "beginner"

    if tone not in valid_tone:
        print("Invalid tone! Default set to 'formal'")
        tone = "formal"

    return audience, tone


# Test
aud, tone = validate_inputs("kids", "cool")
print(aud, tone)
```

```
# Task 6: Prompt Generator

def generate_prompt(topic, audience, tone, style):

    # Step 1: Validate inputs
    audience, tone = validate_inputs(audience, tone)

    # Step 2: Choose template
    if style == "teaching":
        template = "Explain {topic} for {audience} in a {tone} tone step by step"
    elif style == "interview":
        template = "Ask interview questions about {topic} for {audience} in a {tone} tone"
    elif style == "storytelling":
        template = "Explain {topic} for {audience} in a {tone} storytelling style"
    else:
        template = "Explain {topic} for {audience} in a {tone} tone"

    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=template
    )

    # Step 3: Generate prompt
    return prompt.format(topic=topic, audience=audience, tone=tone)


# Test
print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

```
# Task 7: Reusability

reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_inputs = [
    ("AI", "beginner", "fun"),
    ("Python", "expert", "formal"),
    ("ML", "intermediate", "casual"),
    ("Data Science", "beginner", "formal"),
    ("DL", "expert", "technical")  # will fallback tone if invalid
]

for t, a, tone in test_inputs:
    a, tone = validate_inputs(a, tone)
    print(reusable_template.format(topic=t, audience=a, tone=tone))
```

```
# Pipeline Demonstration

def pipeline(topic, audience, tone, style):

    print("User Input Received")

    audience, tone = validate_inputs(audience, tone)
    print("Inputs Validated")

    result = generate_prompt(topic, audience, tone, style)

    print("Prompt Generated:")
    print(result)


pipeline("AI", "beginner", "fun", "teaching")
```
